```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import pandas as pd

if not("titanic" in globals()):
    titanic = pd.read_csv("../data/Titanic-Dataset.csv")

x = titanic[["Pclass", "Sex"]].copy()

y = titanic["Survived"].copy()

x["Sex"] = x["Sex"].map({"male": 0, "female": 1})

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42, stratify=y)

modelo = DecisionTreeClassifier().fit(x_train, y_train)
y_predi = modelo.predict(x_test)

resultados = pd.DataFrame({
    "Real": y_test,
    "Prediction": y_predi
})

# print(resultados.head(15))

# accuracy = (resultados["Real"] == resultados["Prediction"]).mean()
accuracy = accuracy_score(y_test, y_predi)
print(f"Precisión del modelo: {accuracy:.2%}")
```
